```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt

import cartopy.crs as ccrs
import pandas as pd
```

```python
# --- Data Loading (Unchanged) ---
url = 'https://comet.nefsc.noaa.gov/erddap/griddap/noaa_psl_2d74_d418_a6fb'
ds = xr.open_dataset(url)

# --- Longitude Re-indexing (Unchanged) ---
# Convert 0-360 longitude to -180 to 180
ds = ds.assign_coords(longitude = ((ds.longitude + 180) % 360) - 180)
ds = ds.sortby('longitude') 
ds
```

```python
colormap = 'RdBu_r'
date_range = ("1982-01-01", "2024-12-01")

# Northern Hemisphere region 30N to 0, 45W to 0
lat_min = 0
lat_max = 30
lon_min = -45
lon_max = 0

LATITUDE_MIN = lat_min
LATITUDE_MAX = lat_max
LONGITUDE_MIN = lon_min
LONGITUDE_MAX = lon_max

ds_nh = ds['sst'].sel(latitude=slice(LATITUDE_MIN, LATITUDE_MAX), 
                longitude=slice(LONGITUDE_MIN, LONGITUDE_MAX))

# Southern Hemisphere region 20N to 10S, 30W to 15E
lat_min = -10
lat_max = 20
lon_min = -30
lon_max = 15

LATITUDE_MIN = lat_min
LATITUDE_MAX = lat_max
LONGITUDE_MIN = lon_min
LONGITUDE_MAX = lon_max

ds_sh = ds['sst'].sel(latitude=slice(LATITUDE_MIN, LATITUDE_MAX), 
                longitude=slice(LONGITUDE_MIN, LONGITUDE_MAX))
```
