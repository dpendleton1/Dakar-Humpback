```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# --- 1. Create Sample Data ---
# Create a 2D DataArray representing spatial data (e.g., elevation)
lon = np.arange(0, 50, 5)
lat = np.arange(0, 50, 5)
data = np.random.rand(len(lat), len(lon)) * 100

data_array = xr.DataArray(
    data,
    coords=[('lat', lat), ('lon', lon)],
    dims=['lat', 'lon'],
    name='Elevation'
)

# --- 2. Create the Multi-Panel Figure ---

# Use Matplotlib to set up the 2x1 subplot grid
fig, axes = plt.subplots(ncols=2, figsize=(12, 5))

# Plot the first panel (and capture the Axes object)
# xarray's .plot() method returns a dictionary-like object with the Axes
plot_object_1 = data_array.plot(ax=axes[0], cmap='viridis', cbar_kwargs={'label': 'Panel 1 Data'})

# Extract the actual Matplotlib Axes object for the first subplot
ax1 = plot_object_1.axes
ax1.set_title("Panel 1: With Rectangle ROI")


# Plot the second panel
plot_object_2 = data_array.plot(ax=axes[1], cmap='plasma', cbar_kwargs={'label': 'Panel 2 Data'})

# Extract the actual Matplotlib Axes object for the second subplot
ax2 = plot_object_2.axes
ax2.set_title("Panel 2: No Rectangle")


# --- 3. Define and Add the Rectangle Patch to the First Subplot (ax1) ---

# Define the coordinates for the bottom-left corner and the size
lon_start = 10
lat_start = 10
width = 20  # from 10 to 30
height = 15 # from 10 to 25

# Create the Rectangle Patch
rect = patches.Rectangle(
    (lon_start, lat_start), # (x, y) coordinates of the bottom-left corner
    width,                  # width (delta-x)
    height,                 # height (delta-y)
    linewidth=3,
    edgecolor='red',        # Color of the border
    facecolor='none',       # No fill
    linestyle='-'
)

# Add the patch to the first subplot's Axes object
ax1.add_patch(rect)


# --- 4. Final Display/Saving ---
plt.suptitle("Xarray Multi-Panel Plot with Targeted ROI")
plt.tight_layout()
plt.show()
```
