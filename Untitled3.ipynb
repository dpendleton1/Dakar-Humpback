```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# 1. Open NOAA OISST V2 Monthly Mean data via OPeNDAP
# This URL points to the monthly mean NetCDF file
url = "https://psl.noaa.gov/thredds/dodsC/Datasets/noaa.oisst.v2/sst.mnmean.nc"
ds = xr.open_dataset(url)

# 2. Select a subset of time (e.g., 4 months in 2023)
# We select every 3rd month to show seasonal variation
data_to_plot = ds.sst.sel(time=["2023-01-01", "2023-04-01", "2023-07-01", "2023-10-01"])

# 3. Create the Multipanel Plot
# We use 'col' to create subplots based on the 'time' dimension
g = data_to_plot.plot(
    x="lon", y="lat", 
    col="time", col_wrap=2,           # 2x2 grid
    subplot_kw={'projection': ccrs.PlateCarree()}, # Set map projection
    transform=ccrs.PlateCarree(),     # Tell xarray data is in lon/lat
    cmap="RdYlBu_r",                  # Standard temp colormap
    cbar_kwargs={"label": "SST (°C)", "orientation": "horizontal", "shrink": 0.8}
)

# 4. Add Map Features to every subplot
for ax in g.axes.flat:
    ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
    ax.add_feature(cfeature.LAND, facecolor='lightgrey')
    # Add gridlines with labels
    gl = ax.gridlines(draw_labels=True, linestyle='--', alpha=0.5)
    gl.top_labels = False
    gl.right_labels = False

# 5. Fine-tune titles
g.set_titles("{value}") # Simplified title showing only the date
plt.suptitle("NOAA OISST Monthly Mean Sea Surface Temperature (2023)", y=1.02, fontsize=14)

plt.show()
```
